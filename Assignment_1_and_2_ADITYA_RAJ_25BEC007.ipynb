## Question 3: Strings
Write a Python program to accept a string and: count the number of characters, count vowels and consonants, reverse the string, and check whether it is a palindrome.

The program is written as a function and run on two inputs (one palindrome, one non-palindrome). For the palindrome check, case, spaces and punctuation are ignored.

```python
def string_operations(text):
    vowels = "aeiouAEIOU"

    # 1. Number of characters
    total_chars = len(text)
    chars_without_spaces = len(text.replace(" ", ""))

    # 2. Vowels and consonants (only alphabetic characters are counted)
    vowel_count = 0
    consonant_count = 0
    for ch in text:
        if ch.isalpha():
            if ch in vowels:
                vowel_count += 1
            else:
                consonant_count += 1

    # 3. Reverse the string
    reversed_text = text[::-1]

    # 4. Palindrome check (ignore case, spaces and punctuation)
    cleaned = "".join(ch.lower() for ch in text if ch.isalnum())
    is_palindrome = cleaned == cleaned[::-1]

    print(f"\nTotal characters (with spaces)    : {total_chars}")
    print(f"Total characters (without spaces) : {chars_without_spaces}")
    print(f"Number of vowels                  : {vowel_count}")
    print(f"Number of consonants              : {consonant_count}")
    print(f"Reversed string                   : {reversed_text}")
    if is_palindrome:
        print("Palindrome check                  : The string IS a palindrome")
    else:
        print("Palindrome check                  : The string is NOT a palindrome")


# Test case 1 : a palindrome
string_operations(input("Enter a string: "))
```

## Question 5: GPA Calculator
Develop a Python program that accepts the grade/grade point and credit of different courses and calculates the SGPA/GPA using

$$GPA = \frac{\sum (\text{Grade Point} \times \text{Credit})}{\sum \text{Credits}}$$

A letter grade (O, A+, A, B+, B, C, P, F on a 10-point scale) **or** a numeric grade point (0–10) can be entered.

```python
# 10-point grading scale
grade_points = {"O": 10, "A+": 9, "A": 8, "B+": 7, "B": 6, "C": 5, "P": 4, "F": 0}


def get_grade_point(value):
    """Convert a letter grade or numeric grade point to a grade point (None if invalid)."""
    value = value.strip().upper()
    if value in grade_points:
        return grade_points[value]
    try:
        gp = float(value)
        if 0 <= gp <= 10:
            return gp
    except ValueError:
        pass
    return None


num_courses = int(input("Enter the number of courses: "))

records = []
total_weighted_points = 0
total_credits = 0

for i in range(1, num_courses + 1):
    print(f"\nCourse {i}")
    course = input("  Course name                              : ")

    while True:
        gp = get_grade_point(input("  Grade (O/A+/A/B+/B/C/P/F) or grade point : "))
        if gp is not None:
            break
        print("  Invalid grade, please try again.")

    while True:
        try:
            credit = float(input("  Credit                                   : "))
            if credit > 0:
                break
        except ValueError:
            pass
        print("  Invalid credit, please enter a positive number.")

    records.append((course, gp, credit))
    total_weighted_points += gp * credit
    total_credits += credit

gpa = total_weighted_points / total_credits

print("\n" + "=" * 62)
print(f"{'Course':<26}{'Grade Point':>12}{'Credit':>8}{'GP x Credit':>14}")
print("=" * 62)
for course, gp, credit in records:
    print(f"{course:<26}{gp:>12g}{credit:>8g}{gp * credit:>14g}")
print("=" * 62)
print(f"Total credits                 : {total_credits:g}")
print(f"Sum of (Grade Point x Credit) : {total_weighted_points:g}")
print(f"GPA / SGPA                    : {gpa:.2f}")
```

## Question 6: Functions and Iteration
Write separate Python functions to calculate the factorial of a number, generate the Fibonacci series up to *n* terms, and check whether a number is prime.

```python
def factorial(n):
    """Return n! (iterative). Factorial is not defined for negative numbers."""
    if n < 0:
        return None
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result


def fibonacci(n):
    """Return a list with the first n terms of the Fibonacci series."""
    series = []
    a, b = 0, 1
    for _ in range(n):
        series.append(a)
        a, b = b, a + b
    return series


def is_prime(num):
    """Return True if num is a prime number, otherwise False."""
    if num < 2:
        return False
    if num % 2 == 0:
        return num == 2
    divisor = 3
    while divisor * divisor <= num:
        if num % divisor == 0:
            return False
        divisor += 2
    return True


# ----- Using the functions -----
n = int(input("Enter a number to find its factorial: "))
print(f"Factorial of {n} = {factorial(n)}")

terms = int(input("\nEnter the number of Fibonacci terms: "))
print(f"First {terms} Fibonacci terms: {fibonacci(terms)}")

for _ in range(2):
    num = int(input("\nEnter a number to check for prime: "))
    if is_prime(num):
        print(f"{num} is a prime number")
    else:
        print(f"{num} is not a prime number")

print("\nAll prime numbers up to 50:", [x for x in range(1, 51) if is_prime(x)])
```

## Question 10: Mini Programming Problem
Develop a menu-driven Python program for maintaining a student record using a dictionary. Options: **add a student, update marks, search for a student, delete a student, display all student records**.

The dictionary uses the roll number as the key and stores the name and marks of each student. The sample run below exercises every menu option.

```python
students = {}   # roll_no -> {"name": ..., "marks": ...}


def read_marks(prompt):
    """Keep asking until a valid mark between 0 and 100 is entered."""
    while True:
        try:
            marks = float(input(prompt))
            if 0 <= marks <= 100:
                return marks
        except ValueError:
            pass
        print("Invalid marks! Enter a number between 0 and 100.")


def add_student():
    roll = input("Enter roll number : ").strip()
    if roll in students:
        print("A student with this roll number already exists.")
        return
    name = input("Enter name        : ").strip()
    marks = read_marks("Enter marks       : ")
    students[roll] = {"name": name, "marks": marks}
    print(f"Student {name} added successfully.")


def update_marks():
    roll = input("Enter roll number to update marks : ").strip()
    if roll not in students:
        print("Student not found.")
        return
    students[roll]["marks"] = read_marks("Enter new marks : ")
    print(f"Marks updated for {students[roll]['name']}.")


def search_student():
    roll = input("Enter roll number to search : ").strip()
    if roll in students:
        s = students[roll]
        print(f"Student found -> Roll No: {roll} | Name: {s['name']} | Marks: {s['marks']:g}")
    else:
        print("Student not found.")


def delete_student():
    roll = input("Enter roll number to delete : ").strip()
    if roll in students:
        removed = students.pop(roll)
        print(f"Record of {removed['name']} deleted successfully.")
    else:
        print("Student not found.")


def display_all():
    if not students:
        print("No student records available.")
        return
    print(f"\n{'Roll No':<12}{'Name':<20}{'Marks':>8}")
    print("-" * 40)
    for roll, s in students.items():
        print(f"{roll:<12}{s['name']:<20}{s['marks']:>8g}")


# ---------------- Menu loop ----------------
while True:
    print("\n========== STUDENT RECORD MANAGEMENT ==========")
    print("1. Add a student")
    print("2. Update marks")
    print("3. Search for a student")
    print("4. Delete a student")
    print("5. Display all student records")
    print("6. Exit")
    choice = input("Enter your choice (1-6): ").strip()

    if choice == "1":
        add_student()
    elif choice == "2":
        update_marks()
    elif choice == "3":
        search_student()
    elif choice == "4":
        delete_student()
    elif choice == "5":
        display_all()
    elif choice == "6":
        print("Exiting the program. Thank you!")
        break
    else:
        print("Invalid choice! Please select a number from 1 to 6.")
```

## Experimental Setup (common to both lab works)

* **Language / libraries:** Python 3 and the standard library (`time`, `random`, `statistics`, `math`, `gc`); `matplotlib` is used only for plotting.
* **Timer:** `time.perf_counter()` (monotonic, highest available resolution).
* **Repeated trials:** every measurement is repeated several times and the **median** of the trials is reported (medians are less sensitive to occasional OS-scheduling spikes than means). *All reported times in this notebook are medians.*
* **Reproducible inputs:** every input is generated with `random.Random(SEED + n)` using a fixed seed, so the same data is produced on every run.
* **Fair timing:** input generation and validation are done **outside** the timed region; the garbage collector is paused during timing, and in Lab Work 1 the trials are interleaved across input sizes (and a warm-up call is made) so that short bursts of system noise do not distort a single size.

```python
import sys
import platform
import time
import gc
import math
import random
import statistics
import matplotlib.pyplot as plt

SEED = 42

print("Python version :", sys.version.split()[0])
print("Implementation :", platform.python_implementation())
print("Platform       :", platform.platform())
print("Machine        :", platform.machine())
print("Timer          : time.perf_counter(), resolution = %.0e s" % time.get_clock_info("perf_counter").resolution)
print("Random seed    :", SEED)
print("Reported value : MEDIAN of repeated trials")


def rng_for(n):
    """Reproducible random generator for an input of size n."""
    return random.Random(SEED + n)


def time_median(func, args=(), repeats=5, number=1):
    """Median time (seconds) of ONE call of func(*args), over `repeats` trials.
    Each trial runs `number` calls back-to-back and divides by `number`."""
    gc.collect()
    gc_was_on = gc.isenabled()
    gc.disable()
    samples = []
    try:
        for _ in range(repeats):
            start = time.perf_counter()
            for _ in range(number):
                func(*args)
            samples.append((time.perf_counter() - start) / number)
    finally:
        if gc_was_on:
            gc.enable()
    return statistics.median(samples)
```

## Lab Work 1: Benchmarking Algorithmic Growth Rates

**Objective:** Empirically investigate constant, logarithmic, linear, *n* log *n*, quadratic and exponential growth patterns.

### Step 1 — Representative algorithms

| Growth class | Operation / algorithm | Input used (generated outside the timed region) |
|---|---|---|
| Constant, O(1) | list indexing (a batch of 100,000 look-ups) | list `0..n-1`; look-ups cycle through the first, middle and last position |
| Logarithmic, O(log n) | binary search (a batch of 20,000 successful searches) | sorted list of *n* even integers, 20,000 random keys that are present |
| Linear, O(n) | linear search, **worst case** | list `0..n-1`, key `-1` (absent, so all *n* elements are compared) |
| O(n log n) | merge sort | *n* random integers in `[0, 10^6]` |
| Quadratic, O(n²) | double-loop pair counter (pairs with a given sum) | *n* random integers in `[0, 1000]`, target 1000 |
| Exponential, O(2ⁿ) | enumeration of all subsets (count subsets with a given sum) | *n* random integers in `[1, 50]`, target = half of the total |

Single list-index and binary-search operations take far less time than the timer can measure reliably, so they are timed as **batches** of operations.

```python
# ---------- 1. Constant time O(1): list indexing ----------
def index_access(data, indices):
    """Perform len(indices) list look-ups; each look-up is O(1)."""
    for i in indices:
        data[i]


# ---------- 2. Logarithmic time O(log n): binary search ----------
def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1


def binary_search_batch(arr, targets):
    for t in targets:
        binary_search(arr, t)


# ---------- 3. Linear time O(n): linear search ----------
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1


# ---------- 4. O(n log n): merge sort ----------
def merge(left, right):
    merged = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1
    merged.extend(left[i:])
    merged.extend(right[j:])
    return merged


def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    return merge(merge_sort(arr[:mid]), merge_sort(arr[mid:]))


# ---------- 5. Quadratic time O(n^2): double-loop pair counter ----------
def count_pairs(arr, target):
    """Count index pairs (i < j) with arr[i] + arr[j] == target."""
    count = 0
    n = len(arr)
    for i in range(n):
        for j in range(i + 1, n):
            if arr[i] + arr[j] == target:
                count += 1
    return count


# ---------- 6. Exponential time O(2^n): subset enumeration ----------
def count_subsets(arr, target, idx=0, current=0):
    """Visit all 2^n subsets and count those whose sum equals target."""
    if idx == len(arr):
        return 1 if current == target else 0
    return (count_subsets(arr, target, idx + 1, current + arr[idx])   # include arr[idx]
            + count_subsets(arr, target, idx + 1, current))           # exclude arr[idx]


# ---------- Correctness checks (not timed) ----------
rng = random.Random(SEED)
sorted_data = sorted(rng.sample(range(1000), 50))
assert all(binary_search(sorted_data, v) == i for i, v in enumerate(sorted_data))
assert binary_search(sorted_data, -5) == -1
assert linear_search(sorted_data, sorted_data[17]) == 17 and linear_search(sorted_data, -1) == -1
values = [rng.randint(0, 1000) for _ in range(500)]
assert merge_sort(values) == sorted(values)
small = [3, 1, 4, 1, 5, 9, 2, 6]
assert count_pairs(small, 10) == sum(1 for i in range(8) for j in range(i + 1, 8) if small[i] + small[j] == 10)
assert count_subsets([1, 2, 3], 3) == 2            # {3} and {1, 2}
print("All six algorithms passed their correctness checks.")
```

### Step 2 — Measurements
Input sizes are chosen per growth class: very large *n* for the fast classes, small *n* for the quadratic class, and only *n* ≤ 22 for the exponential class.
Each table shows the median time, the **measured ratio** T(n)/T(previous n), the **ideal ratio** g(n)/g(previous n) predicted by the theoretical growth function g, and the **normalised time** T(n)/g(n) (in nanoseconds), which should stay roughly constant if the growth class is right.

```python
results = {}   # name -> dict(n=[...], t=[...] seconds, g=growth function, label=theory)


def benchmark(name, theory, sizes, setup, run, g, repeats):
    """Time run(*setup(n)) for every n in sizes and print a table.
    All inputs are generated first (not timed). The trials are INTERLEAVED across the
    sizes (trial 1 for every n, then trial 2 for every n, ...), so a short burst of
    system noise cannot spoil all trials of one size."""
    inputs = {n: setup(n) for n in sizes}            # input generation: NOT timed
    run(*inputs[sizes[0]])                           # warm-up call (not recorded)
    samples = {n: [] for n in sizes}
    gc.collect()
    gc.disable()
    try:
        for _ in range(repeats):
            for n in sizes:
                start = time.perf_counter()          # timed region starts
                run(*inputs[n])
                samples[n].append(time.perf_counter() - start)
    finally:
        gc.enable()

    print(f"\n{name}   [theory: {theory};  median of {repeats} trials]")
    print(f"{'n':>10} {'time (ms)':>12} {'measured ratio':>15} {'ideal ratio':>12} {'T/g(n) (ns)':>13}")
    ns, ts = [], []
    prev_n = prev_t = None
    for n in sizes:
        t = statistics.median(samples[n])
        if prev_t is None:
            measured, ideal = "-", "-"
        else:
            measured, ideal = f"{t / prev_t:.2f}", f"{g(n) / g(prev_n):.2f}"
        print(f"{n:>10,} {t * 1000:>12.4f} {measured:>15} {ideal:>12} {t * 1e9 / g(n):>13.2f}")
        ns.append(n)
        ts.append(t)
        prev_n, prev_t = n, t
    results[name] = {"n": ns, "t": ts, "g": g, "label": theory}


K_INDEX = 100_000     # look-ups per timed call
K_SEARCH = 20_000     # binary searches per timed call

# ---- 1. List indexing, O(1) ----
def setup_index(n):
    # look-ups cycle through the first, middle and last position of the list
    positions = [0, n // 2, n - 1]
    return list(range(n)), [positions[i % 3] for i in range(K_INDEX)]

benchmark("List indexing", "O(1)", [1_000, 4_000, 16_000, 64_000, 256_000, 1_024_000],
          setup_index, index_access, g=lambda n: K_INDEX, repeats=9)
```

```python
# ---- 2. Binary search, O(log n) ----
def setup_binary(n):
    r = rng_for(n)
    arr = list(range(0, 2 * n, 2))                  # sorted even numbers
    return arr, [r.choice(arr) for _ in range(K_SEARCH)]   # keys that are present

benchmark("Binary search", "O(log n)", [2 ** k for k in range(10, 21)],
          setup_binary, binary_search_batch, g=lambda n: K_SEARCH * math.log2(n), repeats=9)
```

```python
# ---- 3. Linear search (worst case: key absent), O(n) ----
def setup_linear(n):
    return list(range(n)), -1

benchmark("Linear search", "O(n)", [10_000 * 2 ** k for k in range(7)],
          setup_linear, linear_search, g=lambda n: n, repeats=9)
```

```python
# ---- 4. Merge sort, O(n log n) ----
def setup_sort(n):
    r = rng_for(n)
    return ([r.randint(0, 10 ** 6) for _ in range(n)],)

benchmark("Merge sort", "O(n log n)", [1_000 * 2 ** k for k in range(7)],
          setup_sort, merge_sort, g=lambda n: n * math.log2(n), repeats=7)
```

```python
# ---- 5. Double-loop pair counter, O(n^2) ----
def setup_pairs(n):
    r = rng_for(n)
    return [r.randint(0, 1000) for _ in range(n)], 1000

benchmark("Pair counter (double loop)", "O(n^2)", [100 * 2 ** k for k in range(6)],
          setup_pairs, count_pairs, g=lambda n: n * n, repeats=5)
```

```python
# ---- 6. Subset enumeration, O(2^n)  (n is increased by 2, so the ideal ratio is 4) ----
def setup_subsets(n):
    r = rng_for(n)
    arr = [r.randint(1, 50) for _ in range(n)]
    return arr, sum(arr) // 2

benchmark("Subset enumeration", "O(2^n)", list(range(8, 23, 2)),
          setup_subsets, count_subsets, g=lambda n: 2 ** n, repeats=5)
```

**Do the measured trends agree with the expected growth classes?** Yes — each algorithm shows the growth pattern predicted by theory:

* **Quadratic, O(n²) — pair counter:** doubling *n* multiplies the time by roughly 4 (ratios scatter between about 3.5 and 5), the log-log slope is ≈ 2 and only T/n² is flat (T/n grows by more than a factor of 30 over the range).

**Interpreter overhead.** CPython spends tens of nanoseconds on every loop iteration, function call and list operation, so absolute times are far larger than for compiled code. Single list look-ups and binary searches are shorter than what can be timed reliably, which is why they were measured in batches (loop overhead is included and is the same for every *n*). Fixed per-call overhead also hides much of the slow logarithmic growth of binary search.

**Memory-hierarchy effect (not part of the model).** For binary search the normalised curve T/g(n) is flat for small arrays but rises noticeably once the array has more than roughly 5 × 10⁴ elements. Theory counts every primitive operation as equal, but real hardware does not: the most likely explanation is that large arrays no longer fit in the CPU caches, so probes cost more. The list-indexing test touched only three fixed positions to isolate the pure O(1) behaviour from this effect.

Repeated trials, interleaving across sizes, a warm-up call and medians reduce but do not remove this, so conclusions are drawn from the overall trend, the fitted slopes and the normalised curves rather than from any single ratio.

**Conclusion.** The experiments agree with the theoretical classes O(1), O(log n), O(n), O(n log n), O(n²) and O(2ⁿ): the measured trend for each algorithm is the one its theoretical bound predicts, with the deviations explained above (constants, interpreter overhead, caches and noise).

## Lab Work 2: Prefix Averages — Quadratic versus Linear-Time Design

**Objective:** Demonstrate the reduction from quadratic to linear time using a running-sum strategy.

Given a sequence $S$ of $n$ numbers, the *prefix averages* are

$$A[j] = \frac{S[0] + S[1] + \dots + S[j]}{j+1}, \qquad j = 0, 1, \dots, n-1 .$$

* `prefix_average1` — for every $j$ the prefix sum is **recomputed from scratch** with an inner loop.
* `prefix_average2` — a **running sum** is kept and updated once per new element.

### Derivation of operation counts (summations)

Let the primitive operations be *additions* of sequence elements and per-iteration *constant work* (initialising `total`, one division, one store).

**`prefix_average1`** — iteration $j$ of the outer loop performs $j+1$ additions in the inner loop:

$$\text{Additions}_1(n)=\sum_{j=0}^{n-1}(j+1)=\sum_{k=1}^{n}k=\frac{n(n+1)}{2}.$$

If the constant work per outer iteration is $c$ operations, then

$$T_1(n)=\sum_{j=0}^{n-1}\bigl[(j+1)+c\bigr]=\frac{n(n+1)}{2}+cn=\frac{1}{2}n^2+\Bigl(c+\frac12\Bigr)n \;=\;\Theta(n^2).$$

**`prefix_average2`** — iteration $j$ performs exactly one addition plus constant work $c'$ (one division, one store):

$$T_2(n)=\sum_{j=0}^{n-1}\bigl[1+c'\bigr]=(1+c')\,n \;=\;\Theta(n).$$

Auxiliary space is $O(1)$ beyond the output list of size $n$ for both versions. The improved version removes the repeated work because $\text{prefix\_sum}(j) = \text{prefix\_sum}(j-1) + S[j]$.

```python
def prefix_average1(S):
    """Quadratic version: recompute every prefix sum with an inner loop. O(n^2)."""
    n = len(S)
    A = [0] * n
    for j in range(n):
        total = 0
        for i in range(j + 1):          # j + 1 additions in iteration j
            total += S[i]
        A[j] = total / (j + 1)
    return A


def prefix_average2(S):
    """Linear version: keep a running sum, one addition per new element. O(n)."""
    n = len(S)
    A = [0] * n
    total = 0
    for j in range(n):
        total += S[j]                   # exactly one addition per element
        A[j] = total / (j + 1)
    return A


print(prefix_average1([2, 4, 6, 8]), prefix_average2([2, 4, 6, 8]))
```

### Validation tests
Both implementations are checked against an **independent exact oracle** (`fractions.Fraction` arithmetic) for singleton, positive, negative, mixed-value and floating-point sequences. **Empty input:** both functions return an empty list `[]` — the loops never execute, so no division (and therefore no `ZeroDivisionError`) occurs.

```python
from fractions import Fraction


def exact_prefix_averages(S):
    """Oracle using exact rational arithmetic."""
    out, total = [], Fraction(0)
    for j, x in enumerate(S):
        total += Fraction(x)
        out.append(total / (j + 1))
    return out


def matches_oracle(result, S):
    expected = exact_prefix_averages(S)
    return len(result) == len(expected) and all(
        math.isclose(a, float(e), rel_tol=1e-12, abs_tol=1e-12) for a, e in zip(result, expected))


tests = [
    ("empty",     []),
    ("singleton", [5]),
    ("positive",  [1, 2, 3, 4, 5]),
    ("negative",  [-2, -4, -6, -8]),
    ("mixed",     [3, -1, 4, -1, 5, -9, 2, 6]),
    ("floats",    [0.5, 1.5, -2.25, 4.0]),
]

print(f"{'Test':<10} {'Input':<28} {'prefix_average1':<52} {'prefix_average2':<52} {'Result'}")
for label, S in tests:
    r1, r2 = prefix_average1(S), prefix_average2(S)
    ok = (r1 == r2) and matches_oracle(r1, S) and matches_oracle(r2, S)
    show = lambda r: str([round(v, 4) for v in r])
    print(f"{label:<10} {str(S):<28} {show(r1):<52} {show(r2):<52} {'PASS' if ok else 'FAIL'}")
    assert ok

# 200 random sequences (lengths 0..60, mixed signs): both versions must agree exactly
r = random.Random(SEED)
for _ in range(200):
    S = [r.randint(-100, 100) for _ in range(r.randint(0, 60))]
    assert prefix_average1(S) == prefix_average2(S) and matches_oracle(prefix_average2(S), S)
print("\nRandomised check on 200 sequences: both implementations agree with each other and with the exact oracle.")
```

### Operation counts: verifying the summations
The functions are instrumented to count additions and divisions, and the counts are compared with the derived formulas.

```python
def prefix_average1_counted(S):
    n, A, adds, divs = len(S), [0] * len(S), 0, 0
    for j in range(n):
        total = 0
        for i in range(j + 1):
            total += S[i]
            adds += 1
        A[j] = total / (j + 1)
        divs += 1
    return A, adds, divs


def prefix_average2_counted(S):
    n, A, adds, divs = len(S), [0] * len(S), 0, 0
    total = 0
    for j in range(n):
        total += S[j]
        adds += 1
        A[j] = total / (j + 1)
        divs += 1
    return A, adds, divs


print(f"{'n':>6} {'adds (avg1)':>12} {'n(n+1)/2':>10} {'adds (avg2)':>12} {'n':>6} {'divisions (both)':>17}")
for n in [1, 2, 5, 10, 100, 1000]:
    S = list(range(n))
    _, a1, d1 = prefix_average1_counted(S)
    _, a2, d2 = prefix_average2_counted(S)
    assert a1 == n * (n + 1) // 2 and a2 == n and d1 == d2 == n
    print(f"{n:>6} {a1:>12,} {n * (n + 1) // 2:>10,} {a2:>12,} {n:>6} {d1:>17}")
print("\nMeasured counts match the derived formulas exactly.")
```

### Numerical precision
Both algorithms add the elements in the *same order* (`S[0]`, `S[1]`, ...), so floating-point results are in fact **bit-for-bit identical**; the running sum changes the cost, not the rounding behaviour. To study precision, the results are compared with the exact rational answer.

```python
def max_relative_error(result, S):
    worst = 0.0
    for got, exact in zip(result, exact_prefix_averages(S)):
        if exact != 0:
            worst = max(worst, abs(Fraction(got) - exact) / abs(exact))
    return float(worst)


# Case A: many small floating-point numbers (0.1 cannot be represented exactly)
S = [0.1] * 5000
a1, a2 = prefix_average1(S[:1000]), prefix_average2(S[:1000])      # avg1 is O(n^2): keep n small
print("Case A: 5000 copies of 0.1")
print("  avg1 and avg2 bit-for-bit identical on first 1000 values :", a1 == a2)
print(f"  max relative error of prefix_average2 over 5000 values   : {max_relative_error(prefix_average2(S), S):.2e}")

# Case B: cancellation of huge numbers (adding 1.0 to 1e16 is lost because spacing of doubles there is 2.0)
S = [1e16, 1.0, 1.0, 1.0, -1e16]
print("\nCase B: S =", S)
print("  exact averages  :", [float(x) for x in exact_prefix_averages(S)])
print("  prefix_average1 :", prefix_average1(S))
print("  prefix_average2 :", prefix_average2(S))
```

**Observations on numerical precision**

* **The running sum does not change the rounding behaviour.** Both functions add `S[0], S[1], ...` in the same order, so every partial sum is the same floating-point number and the two outputs are identical bit for bit (shown in Case A on the first 1000 values). Speed is gained at no cost in accuracy.
* **Rounding errors accumulate.** In Case A, 0.1 is not exactly representable and the error builds up with the number of additions: after 5000 additions the largest relative error is about 9 × 10⁻¹⁴, roughly 400 times machine epsilon (≈ 2.2 × 10⁻¹⁶). In the worst case the accumulated error bound grows proportionally to *n*.
* **Cancellation can be catastrophic.** In Case B the three 1.0 values are completely absorbed when added to 1e16 (spacing between adjacent doubles there is 2.0), so the final average is 0.0 instead of 0.6. Both implementations fail in the same way, because the problem lies in the order of floating-point additions, not in the loop structure.
* **Remedies.** For integer data Python integers are exact, so the running sum is exact and only the final division rounds. For floating-point data, `math.fsum`, compensated (Kahan) summation or `fractions.Fraction` reduce or remove the error, at some extra cost per element.

**Operation counts (summations).** The derivation $\sum_{j=0}^{n-1}(j+1)=n(n+1)/2$ was confirmed exactly by the instrumented functions: `prefix_average1` performs n(n+1)/2 additions (Θ(n²)) while `prefix_average2` performs exactly n additions (Θ(n)). Both perform n divisions.

**Measured growth versus theory.** When n doubles, the time of `prefix_average1` grows by a factor close to 4 and the time of `prefix_average2` by a factor close to 2 (the printed ratios scatter around these ideal values because of timing noise). The fitted log-log exponents (printed above) are close to 2 and close to 1 respectively, and the measured points follow the fitted c·n² and c·n reference curves. On the much larger sequences (up to 320,000 elements) `prefix_average2` keeps scaling roughly linearly, with about 2× per doubling on average (individual ratios scatter between roughly 1.5 and 3, partly because of memory effects), while the quadratic version would already be impractical: extrapolating from the measured ≈ 0.25–0.3 s at n = 4000, it would need roughly 1.5–2 s at n = 10⁴ and roughly 2–3 minutes at n = 10⁵.

**When does the improved method become measurably faster?**

* The speed-up is modest (roughly 1.2–1.5×) for n = 1–2 (the nested loop has some extra set-up cost), reaches about 2× at n ≈ 3–5 and roughly 10× for n of about 20–50 (these small-n timings are the noisiest).
* For tiny sequences the absolute difference is only a few microseconds, so it is *not noticeable in practice*; the difference becomes visible in absolute terms (milliseconds) from a few hundred elements onwards.
* The speed-up keeps growing roughly in proportion to n (roughly 60–70× at n = 250 and roughly 800–1000× at n = 4000 here), because the ratio of T₁ ≈ ½·n² to T₂ ≈ c·n is proportional to n.

**Empty input and correctness.** Empty input returns an empty list from both functions with no exception. For singleton, positive, negative, mixed and floating-point sequences (and 200 random sequences) both functions agree exactly and match an exact rational-arithmetic oracle.

**Conclusion.** Maintaining a running sum reduces the running time from Θ(n²) to Θ(n) — each new prefix sum is the previous one plus one new element, so earlier work is reused instead of repeated — without changing the numerical results.
